```python
import numpy as np
from astropy.table import Table

def sin(x):
    return np.sin(x)
```

# This function returns the sine of x

```python
def cos(x):
    return np.cos(x)
```

# This function returns the cosine of x

```python
x = np.linspace(0, 2*np.pi, 1000)
y = sin(x)
z = cos(x)
```

```python
t = Table([x, y, z], names=["X", "Sin(X)", "Cos(x)"])
print(t)
```
