```python
import numpy as np
import matplotlib.pyplot as plt

a = 1
b = 1
x = np.linspace(0, 2 * np.pi, 100)


def f(x):
    return a * np.exp(-b * x)
```

```python
y = f(x)

plt.figure(figsize=(5, 5))
plt.plot(x, y)

plt.xlabel('x')
plt.ylabel('y')

plt.savefig('journal1output.pdf')
plt.show()
```
